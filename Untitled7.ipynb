```python
def load_image(image_path):
  img_bgr=cv2.imread(image_path)
  img_rgb=cv2.cvtColorcolor(img_bgr,cv2.Color_BRG2RGB)
  return img_bgr,img_rgb
```

```python
def preprocess(img_bgr):
  gray=cv2.cvtColorcolor(img_bgr,cv2.COLOR_BQR2GRAY)

  resized=cv2.resize(gray,(640,640))
  return gray,resized
```

```python
def filtering(gray):
  guassian=cv2.GaussianBlur(gray,(5,5),0)

  median=cv2.medianBlur(gray,5)

  bilateral=cv2.bilateralFilter(gray,9,75,75)
  return gaussian,median,bilateral
```

```python
def edge_detection(filtered):
  canny = cv2.Canny(filtered, 100, 200)

  sobelx=cv2.Sobel(filtered,cv2.CV_64F,1,0,ksize=5)
  sobely=cv2.Sobel(filtered,cv2.CV_64F,0,1,ksize=5)
  sobel=cv2.magnitude(sobelx,sobely)
  return canny,sobel
```

```python
def segmentation(gray):

  thresh=cv2.threshold(gray,127,255,cv2.THRESH_BINARY)

  adaptive-cv2.adaptiveTheshold(gray,255,cv2.ADAPTIVE_THRESHOLD_GAUSSIAN_C,cv2.THRES_BINARY,11,2)
  return thresh,adaptive
```

```python
def show_results(original,gray,gaussian,median,edges,thresh):

  plt.figure(figsize=(15,10))

  plt.subplot(2,3,1)
  plt.title("original")
  plt.imshow(original)
  plt.axis("off")


  plt.subplot(2,3,2)
  plt.title("Gray")
  plt.imshow(gray,cmap='gray')
  plt.axis("off")


  plt.subplot(2,3,3)
  plt.title("Gaussian filter")
  plt.imshow(gaussian,cmap='gray')
  plt.axis("off")


  plt.subplot(2,3,4)
  plt.title("Median filter")
  plt.imshow(median,cmap='gray')
  plt.axis("off")


  plt.subplot(2,3,5)
  plt.title("Canny edge detection")
  plt.imshow(edges,cmap='gray')
  plt.axis("off")


  plt.subplot(2,3,6)
  plt.title("Threshold segmentation")
  plt.imshow(thresh,cap='gray')
  plt.axis("off")

  plt.show()
```
